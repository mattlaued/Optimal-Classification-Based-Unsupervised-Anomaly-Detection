````python
from transformers import AutoImageProcessor, AutoModel
from PIL import Image
from transformers import ViTFeatureExtractor, ViTModel, ViTMAEForPreTraining
from transformers import CLIPProcessor, CLIPModel
import os
import numpy as np
# import requests
````

````python
# Config
# DATASETS = ['transistor', 'tile', 'screw', 'grid', 'cable']
DATASETS = ['metal_nut', 'carpet']
MODELS = {# model of similar size (1.22 GB)
    'dinov2': ('facebook/dinov2-large', ['cls', 'mean'], True),
    'clip': ('openai/clip-vit-large-patch14', ['cls'], True),
    'mae': ('facebook/vit-mae-large', ['mean'], False),
    'vit': ('google/vit-large-patch16-224-in21k', ['cls', 'mean'], True)
    }
````

````python
from vision_foundation import get_path, cal_emb
````

````python
# DINOv2
cur_model = 'dinov2'
hf_name, agg, cls = MODELS[cur_model]

processor = AutoImageProcessor.from_pretrained(hf_name)
model = AutoModel.from_pretrained(hf_name)
for dataset in DATASETS:
    print(dataset)
    base_path = f'./data/{dataset}'
    categories, path_dict = get_path(base_path)
    cal_emb(
        processor,
        model,
        model_type=cur_model,
        path_dict=path_dict,
        base_path=base_path,
        agg=agg,
        cls=cls
    )
````

## How to use the results?

**1. Folder**
```python
f'./data/{dataset}/embeddings'
```

**2. File name:**

`dinov2-mean-test-cut_lead-004.npy`

Or the abstract form:

`{model_type}-{agg_method}-{split}-{class}-{file_number}.npy`

Options:
- `dataset`: `['transistor', 'tile', 'screw', 'grid', 'cable']`
- `model_type`: `["dinov2","clip", "mae", "vit"]`
- `agg_method`: `["cls", "mean"]`
- `split`: `["train", "test"]`
- `file_number`: Original file number in the dataset
